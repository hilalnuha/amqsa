````python

# STEP B3 — Definitions: SDS-FDR + fair-budget solvers
# Cell ini hanya mendefinisikan model/solver. Belum menjalankan 20-run experiment.

import time
import numpy as np
import pandas as pd
from scipy.stats import wilcoxon, friedmanchisquare, rankdata

# -------------------------- RUN FLAGS --------------------------
RUN_CGK       = True
RUN_ABLATION  = True
RUN_BENCHMARK = True

# Optional visual analyses from the old notebook are intentionally omitted here.
# Phase 2 focuses on budget fairness + authoritative numerical outputs.

# -------------------------- FAIR BUDGET ------------------------
POP = 50
MAX_EVALS = 5_000
N_RUNS = 20

BD = 30
BPOP = 30
BENCH_MAX_EVALS = 10_000
BN_RUNS = 20

P_CANCEL_INIT = 0.05
QUENCH_FRAC = 0.60

CGK_GRID = np.arange(50, MAX_EVALS + 1, 50)       # 100 checkpoints
BENCH_GRID = np.linspace(100, BENCH_MAX_EVALS, 100).astype(int)

# -------------------------- INSTANCE ---------------------------
# Freeze the same semi-synthetic CGK bank used in the submitted experiment.
# A real/OpenSky disruption instance will be treated as a separate validation experiment.
USE_OPENSKY_FILE = False

RNG_DATA = np.random.default_rng(7)

REOPEN = 180.0
RECOVERY_CUTOFF = 260.0   # 22:20 experimental recovery cutoff/horizon
CP = 6_000.0
COMP_CANCEL = 300_000.0
REV_LOSS_FRAC = 0.7
PAX_CANCEL_LOSS = 1_300_000.0
CANCEL_EQUIV_MIN = 600.0

FLEET = {
    "B738":  (189, "M", "jet",  25_000, 60e6,  "dom"),
    "B739":  (215, "M", "jet",  27_000, 60e6,  "dom"),
    "A320":  (180, "M", "jet",  24_000, 60e6,  "dom"),
    "A333":  (290, "H", "jet",  65_000, 200e6, "intl"),
    "B77W":  (393, "H", "jet",  75_000, 250e6, "intl"),
    "ATR72": (72,  "M", "prop", 10_000, 30e6,  "reg"),
}
FARE = {"dom": (1.0e6, 1.5e6), "intl": (2.6e6, 3.8e6), "reg": (0.7e6, 1.1e6)}

OPERATORS = [
    ("GIA", "Garuda",        {"B738": 0.5, "A333": 0.3, "B77W": 0.2}),
    ("LNI", "Lion Air",      {"B738": 0.55, "B739": 0.45}),
    ("CTV", "Citilink",      {"A320": 1.0}),
    ("BTK", "Batik Air",     {"A320": 0.6, "B738": 0.4}),
    ("SJV", "Super Air Jet", {"A320": 1.0}),
    ("WON", "Wings Air",     {"ATR72": 1.0}),
    ("AWQ", "AirAsia Indonesia", {"A320": 1.0}),
    ("INTL", "Foreign carrier",  {"B77W": 0.6, "A333": 0.4}),
]
OP_WEIGHTS = [0.16, 0.28, 0.14, 0.12, 0.10, 0.06, 0.06, 0.08]
PREFIX2OP = {p: (name, fleet) for p, name, fleet in OPERATORS}
SIDS = ["CKG1A", "DOLTA1", "PURWO1", "NOKTA1"]


def _flight_row(i, prefix, airline, ac, sched):
    cap, wake, eng, cfm, repos, band = FLEET[ac]
    lo, hi = FARE[band]
    return {
        "flight": f"{prefix}{RNG_DATA.integers(100, 999)}",
        "airline": airline,
        "aircraft": ac,
        "capacity": cap,
        "wake": wake,
        "engine": eng,
        "sid": SIDS[int(RNG_DATA.integers(0, len(SIDS)))],
        "sched_min_from_1800": sched,
        "cfm_idr_per_min": cfm,
        "reposition_idr": repos,
        "fare_idr": float(np.round(RNG_DATA.uniform(lo, hi), -3)),
    }


def build_frozen_dataset():
    n = 48
    sched = np.sort(RNG_DATA.choice(np.arange(1, 180), size=n, replace=False))
    rows = []
    for i in range(n):
        k = RNG_DATA.choice(len(OPERATORS), p=OP_WEIGHTS)
        prefix, airline, fleet = OPERATORS[k]
        ac = RNG_DATA.choice(list(fleet), p=list(fleet.values()))
        rows.append(_flight_row(i, prefix, airline, ac, float(sched[i])))
    return pd.DataFrame(rows), "seeded semi-synthetic CGK evening bank (seed=7)"


FLIGHTS, DATA_SOURCE = build_frozen_dataset()
FLIGHTS.to_csv(OUTPUT_DIR / "phase2_cgk_instance.csv", index=False)

N = len(FLIGHTS)
SCHED = FLIGHTS["sched_min_from_1800"].to_numpy(float)
CAP = FLIGHTS["capacity"].to_numpy(float)
NP_SEATS = 0.75 * CAP
CFM = FLIGHTS["cfm_idr_per_min"].to_numpy(float)
FARE_V = FLIGHTS["fare_idr"].to_numpy(float)
REPOS = FLIGHTS["reposition_idr"].to_numpy(float)
WAKE = FLIGHTS["wake"].to_numpy()
ENG = FLIGHTS["engine"].to_numpy()
SID = FLIGHTS["sid"].to_numpy()


def _sep_rule(prev, nxt):
    s = 120.0 if SID[prev] == SID[nxt] else 90.0
    if WAKE[prev] == "H" and WAKE[nxt] == "M":
        s = max(s, 120.0)
    if ENG[prev] == "prop" and ENG[nxt] == "jet" and SID[prev] == SID[nxt]:
        s = max(s, 150.0)
    return s / 60.0


SEP_M = np.array([[_sep_rule(u, v) for v in range(N)] for u in range(N)])


def separation(prev, nxt):
    return SEP_M[prev, nxt]


def mu_disappointment(t):
    return np.clip(np.cbrt((t / 29.0) ** 2 / 29.0), 0.0, 1.0)


def decode(perm, cancel):
    f1 = f2 = f3 = 0.0
    delays = np.full(N, -1.0)
    prev = -1
    tcur = REOPEN
    ncan = 0

    for fl in perm:
        if cancel[fl]:
            f1 += (COMP_CANCEL + REV_LOSS_FRAC * FARE_V[fl]) * NP_SEATS[fl] + REPOS[fl]
            f2 += PAX_CANCEL_LOSS * NP_SEATS[fl]
            f3 += CANCEL_EQUIV_MIN
            ncan += 1
            continue

        dep = tcur if prev < 0 else tcur + separation(prev, fl)

        if dep > RECOVERY_CUTOFF:
            f1 += (COMP_CANCEL + REV_LOSS_FRAC * FARE_V[fl]) * NP_SEATS[fl] + REPOS[fl]
            f2 += PAX_CANCEL_LOSS * NP_SEATS[fl]
            f3 += CANCEL_EQUIV_MIN
            ncan += 1
            continue

        t = dep - SCHED[fl]
        delays[fl] = t
        f1 += t * CFM[fl] + mu_disappointment(t) * NP_SEATS[fl] * FARE_V[fl]
        f2 += NP_SEATS[fl] * CP * t
        f3 += t
        prev, tcur = fl, dep

    return f1, f2, f3, ncan, delays


# Freeze the same random-schedule normalization protocol.
_r = np.random.default_rng(0)
_samples = []
for _ in range(300):
    p = _r.permutation(N)
    c = _r.random(N) < 0.05
    _samples.append(decode(p, c)[:3])

F_REF = np.array(_samples).mean(axis=0)
W = np.array([1/3, 1/3, 1/3])


def fitness(perm, cancel):
    f1, f2, f3, ncan, _ = decode(perm, cancel)
    f = np.array([f1, f2, f3])
    return float(np.sum(W * f / F_REF)), f, ncan


# Priority operator from the submitted implementation.
WGT_PRI = NP_SEATS * CP + CFM + NP_SEATS * FARE_V / 300.0


# --------------------- BUDGET / CURVE RECORDER ---------------------
class CGKBudget:
    def __init__(self, max_evals=MAX_EVALS, grid=CGK_GRID):
        self.max_evals = int(max_evals)
        self.evals = 0
        self.grid = np.asarray(grid, int)
        self.points = []
        self.gi = 0

    @property
    def remaining(self):
        return self.max_evals - self.evals

    def evaluate(self, perm, cancel):
        if self.evals >= self.max_evals:
            raise RuntimeError("Evaluation budget exhausted")
        out = fitness(perm, cancel)
        self.evals += 1
        return out

    def observe(self, best):
        best_f1 = float(best[2][1][0])
        while self.gi < len(self.grid) and self.evals >= self.grid[self.gi]:
            self.points.append(best_f1)
            self.gi += 1

    def curve(self, best):
        self.observe(best)
        while len(self.points) < len(self.grid):
            self.points.append(float(best[2][1][0]))
        return np.asarray(self.points, float)


def _init_perm_pop(rng):
    return ([rng.permutation(N) for _ in range(POP)],
            [rng.random(N) < P_CANCEL_INIT for _ in range(POP)])


def _ox(rng, p1, p2):
    a, b = sorted(rng.choice(N, 2, replace=False))
    child = -np.ones(N, int)
    child[a:b] = p1[a:b]
    seg = set(p1[a:b])
    fill = [g for g in p2 if g not in seg]
    child[child < 0] = fill
    return child


def _update_best(best, p, c, fit):
    if best is None or fit[0] < best[2][0]:
        return (p.copy(), c.copy(), fit)
    return best


# -------------------------- FAIR CGK SOLVERS --------------------------
def run_ga(seed, max_evals=MAX_EVALS):
    rng = np.random.default_rng(seed)
    pc, pm = 1.0, 0.2
    budget = CGKBudget(max_evals)
    perms, cans = _init_perm_pop(rng)

    fits, best = [], None
    for p, c in zip(perms, cans):
        f = budget.evaluate(p, c)
        fits.append(f)
        best = _update_best(best, p, c, f)
        budget.observe(best)

    while budget.remaining >= POP:
        np_perms, np_cans, np_fits = [], [], []
        while len(np_perms) < POP:
            i1 = min(rng.choice(POP, 3, replace=False), key=lambda i: fits[i][0])
            i2 = min(rng.choice(POP, 3, replace=False), key=lambda i: fits[i][0])

            if rng.random() < pc:
                child = _ox(rng, perms[i1], perms[i2])
                mask = rng.random(N) < 0.5
                cbit = np.where(mask, cans[i1], cans[i2])
            else:
                child, cbit = perms[i1].copy(), cans[i1].copy()

            if rng.random() < pm:
                a, b = rng.choice(N, 2, replace=False)
                child[a], child[b] = child[b], child[a]

            if rng.random() < 0.1:
                k = rng.integers(0, N)
                cbit = cbit.copy()
                cbit[k] = ~cbit[k]

            f = budget.evaluate(child, cbit)
            np_perms.append(child)
            np_cans.append(cbit)
            np_fits.append(f)
            best = _update_best(best, child, cbit, f)
            budget.observe(best)

        perms, cans, fits = np_perms, np_cans, np_fits
        wi = int(np.argmax([f[0] for f in fits]))
        perms[wi], cans[wi], fits[wi] = best[0].copy(), best[1].copy(), best[2]

    assert budget.evals == max_evals
    return best, budget.curve(best), budget.evals


def run_sa(seed, max_evals=MAX_EVALS):
    rng = np.random.default_rng(seed)
    budget = CGKBudget(max_evals)

    cur_p = rng.permutation(N)
    cur_c = rng.random(N) < P_CANCEL_INIT
    cur_f = budget.evaluate(cur_p, cur_c)
    best = (cur_p.copy(), cur_c.copy(), cur_f)
    budget.observe(best)

    T, alpha = 0.02, 0.999

    while budget.remaining > 0:
        cp, cc = cur_p.copy(), cur_c.copy()
        u = rng.random()

        if u < 0.45:
            a, b = rng.choice(N, 2, replace=False)
            cp[a], cp[b] = cp[b], cp[a]
        elif u < 0.90:
            a, b = rng.choice(N, 2, replace=False)
            g = cp[a]
            cp = np.delete(cp, a)
            cp = np.insert(cp, b, g)
        else:
            k = rng.integers(0, N)
            cc[k] = ~cc[k]

        cf = budget.evaluate(cp, cc)
        d = cf[0] - cur_f[0]

        if d < 0 or rng.random() < np.exp(-d / max(T, 1e-12)):
            cur_p, cur_c, cur_f = cp, cc, cf
            best = _update_best(best, cp, cc, cf)

        T *= alpha
        budget.observe(best)

    assert budget.evals == max_evals
    return best, budget.curve(best), budget.evals


def _keys_decode(keys):
    return np.argsort(keys[:N], kind="stable"), keys[N:] > 0.9


def run_pso(seed, max_evals=MAX_EVALS):
    rng = np.random.default_rng(seed)
    D = 2 * N
    budget = CGKBudget(max_evals)

    x = rng.random((POP, D))
    v = np.zeros_like(x)
    pb = x.copy()

    pf, best = [], None
    for xi in x:
        p, c = _keys_decode(xi)
        f = budget.evaluate(p, c)
        pf.append(f)
        best = _update_best(best, p, c, f)
        budget.observe(best)

    gi = int(np.argmin([f[0] for f in pf]))
    gb, gf = pb[gi].copy(), pf[gi]

    total_iters = (max_evals - POP) // POP
    it = 0

    while budget.remaining >= POP:
        frac = it / max(total_iters - 1, 1)
        w = 0.9 - 0.5 * frac
        r1, r2 = rng.random((POP, D)), rng.random((POP, D))
        v = np.clip(w * v + 1.8 * r1 * (pb - x) + 1.8 * r2 * (gb - x), -0.5, 0.5)
        x = np.clip(x + v, 0, 1)

        for i in range(POP):
            p, c = _keys_decode(x[i])
            fi = budget.evaluate(p, c)

            if fi[0] < pf[i][0]:
                pb[i], pf[i] = x[i].copy(), fi
                if fi[0] < gf[0]:
                    gb, gf = x[i].copy(), fi

            best = _update_best(best, p, c, fi)
            budget.observe(best)

        it += 1

    assert budget.evals == max_evals
    p, c = _keys_decode(gb)
    solver_best = (p, c, gf)
    # 'best' and solver_best should agree on scalar objective.
    if best[2][0] < solver_best[2][0]:
        solver_best = best
    return solver_best, budget.curve(solver_best), budget.evals


def run_zoa(seed, max_evals=MAX_EVALS):
    rng = np.random.default_rng(seed)
    D = 2 * N
    budget = CGKBudget(max_evals)

    x = rng.random((POP, D))
    fits, best = [], None
    for xi in x:
        p, c = _keys_decode(xi)
        f = budget.evaluate(p, c)
        fits.append(f)
        best = _update_best(best, p, c, f)
        budget.observe(best)

    gi = int(np.argmin([f[0] for f in fits]))
    pz, pzf = x[gi].copy(), fits[gi]

    while budget.remaining > 0:
        for i in range(POP):
            if budget.remaining <= 0:
                break

            # Phase 1: foraging
            r = rng.random(D)
            I = 1 + rng.integers(0, 2)
            xn = np.clip(x[i] + r * (pz - I * x[i]), 0, 1)
            p, c = _keys_decode(xn)
            fn = budget.evaluate(p, c)

            if fn[0] < fits[i][0]:
                x[i], fits[i] = xn, fn
            if fits[i][0] < pzf[0]:
                pz, pzf = x[i].copy(), fits[i]

            bp, bc = _keys_decode(pz)
            best = _update_best(best, bp, bc, pzf)
            budget.observe(best)

            if budget.remaining <= 0:
                break

            # Phase 2: defense; schedule based on evaluation progress.
            progress = budget.evals / max_evals
            if rng.random() < 0.5:
                xn = np.clip(
                    x[i] + 0.01 * (2 * rng.random(D) - 1) * (1 - progress) * x[i],
                    0, 1
                )
            else:
                k = rng.integers(0, POP)
                r = rng.random(D)
                I = 1 + rng.integers(0, 2)
                xn = np.clip(x[i] + r * (x[k] - I * x[i]), 0, 1)

            p, c = _keys_decode(xn)
            fn = budget.evaluate(p, c)

            if fn[0] < fits[i][0]:
                x[i], fits[i] = xn, fn
            if fits[i][0] < pzf[0]:
                pz, pzf = x[i].copy(), fits[i]

            bp, bc = _keys_decode(pz)
            best = _update_best(best, bp, bc, pzf)
            budget.observe(best)

    assert budget.evals == max_evals
    p, c = _keys_decode(pz)
    solver_best = (p, c, pzf)
    if best[2][0] < solver_best[2][0]:
        solver_best = best
    return solver_best, budget.curve(solver_best), budget.evals


def _order_impose(rng, perm, guide, frac):
    s = max(2, int(np.ceil(max(frac, 1e-9) * N)))
    s = min(s, N)
    subset = rng.choice(N, s, replace=False)
    mask = np.isin(perm, subset)
    pos_g = np.empty(N, int)
    pos_g[guide] = np.arange(N)
    ordered = sorted(perm[mask], key=lambda f: pos_g[f])
    out = perm.copy()
    out[mask] = ordered
    return out


def run_dzoa(seed, max_evals=MAX_EVALS):
    rng = np.random.default_rng(seed)
    budget = CGKBudget(max_evals)
    perms, cans = _init_perm_pop(rng)

    fits, best = [], None
    for p, c in zip(perms, cans):
        f = budget.evaluate(p, c)
        fits.append(f)
        best = _update_best(best, p, c, f)
        budget.observe(best)

    gi = int(np.argmin([f[0] for f in fits]))
    pz_p, pz_c, pz_f = perms[gi].copy(), cans[gi].copy(), fits[gi]

    while budget.remaining > 0:
        for i in range(POP):
            if budget.remaining <= 0:
                break

            # Phase 1
            r = rng.random()
            I = 1 + rng.integers(0, 2)
            xp = _order_impose(rng, perms[i], pz_p, frac=0.25 * r * I)
            xc = cans[i].copy()
            m = rng.random(N) < 0.5 * r
            xc = np.where(m, pz_c, xc)

            fn = budget.evaluate(xp, xc)
            if fn[0] < fits[i][0]:
                perms[i], cans[i], fits[i] = xp, xc, fn
            if fits[i][0] < pz_f[0]:
                pz_p, pz_c, pz_f = perms[i].copy(), cans[i].copy(), fits[i]
            best = _update_best(best, pz_p, pz_c, pz_f)
            budget.observe(best)

            if budget.remaining <= 0:
                break

            # Phase 2
            progress = budget.evals / max_evals
            if rng.random() < 0.5:
                kswap = max(1, int(round(4 * (1 - progress))))
                xp = perms[i].copy()
                for _ in range(kswap):
                    a, b = rng.choice(N, 2, replace=False)
                    xp[a], xp[b] = xp[b], xp[a]
                xc = cans[i].copy()
                if rng.random() < 0.2:
                    j = rng.integers(0, N)
                    xc[j] = ~xc[j]
            else:
                k = rng.integers(0, POP)
                r = rng.random()
                I = 1 + rng.integers(0, 2)
                xp = _order_impose(rng, perms[i], perms[k], frac=0.25 * r * I)
                xc = cans[i].copy()
                m = rng.random(N) < 0.5 * r
                xc = np.where(m, cans[k], xc)

            fn = budget.evaluate(xp, xc)
            if fn[0] < fits[i][0]:
                perms[i], cans[i], fits[i] = xp, xc, fn
            if fits[i][0] < pz_f[0]:
                pz_p, pz_c, pz_f = perms[i].copy(), cans[i].copy(), fits[i]
            best = _update_best(best, pz_p, pz_c, pz_f)
            budget.observe(best)

    assert budget.evals == max_evals
    solver_best = (pz_p, pz_c, pz_f)
    if best[2][0] < solver_best[2][0]:
        solver_best = best
    return solver_best, budget.curve(solver_best), budget.evals


def run_amqsa(seed, max_evals=MAX_EVALS, T0=0.02, alpha=0.999,
              lam=0.9, quench_frac=QUENCH_FRAC):
    rng = np.random.default_rng(seed)
    budget = CGKBudget(max_evals)

    cur_p = rng.permutation(N)
    cur_c = rng.random(N) < P_CANCEL_INIT
    cur_f = budget.evaluate(cur_p, cur_c)
    best = (cur_p.copy(), cur_c.copy(), cur_f)
    budget.observe(best)

    T = T0
    wts = np.array([1.5, 1.5, 1.0, 1.0, 0.3])
    quenched = False

    while budget.remaining > 0:
        if not quenched and budget.evals >= quench_frac * max_evals:
            T = 1e-12
            quenched = True
            cur_p, cur_c, cur_f = best[0].copy(), best[1].copy(), best[2]

        pr = np.maximum(wts, 0.05)
        pr = pr / pr.sum()
        oi = rng.choice(5, p=pr)

        cp, cc = cur_p.copy(), cur_c.copy()

        if oi == 0:
            a, b = rng.choice(N, 2, replace=False)
            cp[a], cp[b] = cp[b], cp[a]
        elif oi == 1:
            a, b = rng.choice(N, 2, replace=False)
            g = cp[a]
            cp = np.delete(cp, a)
            cp = np.insert(cp, b, g)
        elif oi == 2:
            L = int(rng.integers(2, 4))
            a = int(rng.integers(0, N - L))
            blk = cp[a:a+L].copy()
            rest = np.concatenate([cp[:a], cp[a+L:]])
            b = int(rng.integers(0, len(rest) + 1))
            cp = np.concatenate([rest[:b], blk, rest[b:]])
        elif oi == 3:
            i, j = sorted(rng.choice(N, 2, replace=False))
            if WGT_PRI[cp[j]] <= WGT_PRI[cp[i]]:
                i, j = j, i
            cp[i], cp[j] = cp[j], cp[i]
        else:
            k = rng.integers(0, N)
            cc[k] = ~cc[k]

        cf = budget.evaluate(cp, cc)
        d = cf[0] - cur_f[0]
        acc = d < 0 or rng.random() < np.exp(-d / max(T, 1e-12))

        wts[oi] = lam * wts[oi] + (1-lam) * (1.0 if d < 0 else (0.3 if acc else 0.0))

        if acc:
            cur_p, cur_c, cur_f = cp, cc, cf
            best = _update_best(best, cp, cc, cf)

        T *= alpha
        budget.observe(best)

    assert budget.evals == max_evals
    return best, budget.curve(best), budget.evals


ALGOS = {
    "GA": run_ga,
    "SA": run_sa,
    "PSO": run_pso,
    "ZOA": run_zoa,
    "dZOA": run_dzoa,
    "AMQSA": run_amqsa,
}


# -------------------------- ABLATION FAIR BUDGET --------------------------
def _ablation_variant(seed, ops, w0, max_evals=MAX_EVALS,
                      T0=0.02, alpha=0.999, lam=0.9,
                      quench=QUENCH_FRAC, reheat=False,
                      stag_limit=400, rf=25.0):
    rng = np.random.default_rng(seed)
    budget = CGKBudget(max_evals)

    cur_p = rng.permutation(N)
    cur_c = rng.random(N) < P_CANCEL_INIT
    cur_f = budget.evaluate(cur_p, cur_c)
    best = (cur_p.copy(), cur_c.copy(), cur_f)

    T = T0
    wts = np.array(w0, float)
    quenched = False
    stag = 0

    while budget.remaining > 0:
        if not reheat and not quenched and budget.evals >= quench * max_evals:
            T = 1e-12
            quenched = True
            cur_p, cur_c, cur_f = best[0].copy(), best[1].copy(), best[2]

        pr = np.maximum(wts, 0.05)
        pr = pr / pr.sum()
        oi = rng.choice(len(ops), p=pr)
        op = ops[oi]

        cp, cc = cur_p.copy(), cur_c.copy()

        if op == "swap":
            a, b = rng.choice(N, 2, replace=False)
            cp[a], cp[b] = cp[b], cp[a]
        elif op == "ins":
            a, b = rng.choice(N, 2, replace=False)
            g = cp[a]
            cp = np.delete(cp, a)
            cp = np.insert(cp, b, g)
        elif op == "blk":
            L = int(rng.integers(2, 4))
            a = int(rng.integers(0, N - L))
            blk = cp[a:a+L].copy()
            rest = np.concatenate([cp[:a], cp[a+L:]])
            b = int(rng.integers(0, len(rest) + 1))
            cp = np.concatenate([rest[:b], blk, rest[b:]])
        elif op == "pri":
            i, j = sorted(rng.choice(N, 2, replace=False))
            if WGT_PRI[cp[j]] <= WGT_PRI[cp[i]]:
                i, j = j, i
            cp[i], cp[j] = cp[j], cp[i]
        else:
            k = rng.integers(0, N)
            cc[k] = ~cc[k]

        cf = budget.evaluate(cp, cc)
        d = cf[0] - cur_f[0]
        acc = d < 0 or rng.random() < np.exp(-d / max(T, 1e-12))

        wts[oi] = lam*wts[oi] + (1-lam)*(1.0 if d < 0 else (0.3 if acc else 0.0))

        if acc:
            cur_p, cur_c, cur_f = cp, cc, cf
            if cf[0] < best[2][0]:
                best = (cp.copy(), cc.copy(), cf)
                stag = 0
            else:
                stag += 1
        else:
            stag += 1

        if reheat and stag >= stag_limit:
            T = min(T0, T * rf)
            cur_p, cur_c, cur_f = best[0].copy(), best[1].copy(), best[2]
            stag = 0

        T *= alpha

    return float(best[2][1][0]), budget.evals


def _quench_only(seed, max_evals=MAX_EVALS, T0=0.02, alpha=0.999,
                 quench_frac=QUENCH_FRAC):
    rng = np.random.default_rng(seed)
    budget = CGKBudget(max_evals)

    cur_p = rng.permutation(N)
    cur_c = rng.random(N) < P_CANCEL_INIT
    cur_f = budget.evaluate(cur_p, cur_c)
    best = (cur_p.copy(), cur_c.copy(), cur_f)

    T = T0
    quenched = False

    while budget.remaining > 0:
        if not quenched and budget.evals >= quench_frac * max_evals:
            T = 1e-12
            quenched = True
            cur_p, cur_c, cur_f = best[0].copy(), best[1].copy(), best[2]

        cp, cc = cur_p.copy(), cur_c.copy()
        u = rng.random()

        if u < 0.45:
            a, b = rng.choice(N, 2, replace=False)
            cp[a], cp[b] = cp[b], cp[a]
        elif u < 0.90:
            a, b = rng.choice(N, 2, replace=False)
            g = cp[a]
            cp = np.delete(cp, a)
            cp = np.insert(cp, b, g)
        else:
            k = rng.integers(0, N)
            cc[k] = ~cc[k]

        cf = budget.evaluate(cp, cc)
        d = cf[0] - cur_f[0]

        if d < 0 or rng.random() < np.exp(-d / max(T, 1e-12)):
            cur_p, cur_c, cur_f = cp, cc, cf
            best = _update_best(best, cp, cc, cf)

        T *= alpha

    return float(best[2][1][0]), budget.evals


# -------------------------- BENCHMARK FUNCTIONS --------------------------
def sphere(x):
    return float(np.sum(x*x))

def rosenbrock(x):
    return float(np.sum(100*(x[1:] - x[:-1]**2)**2 + (1-x[:-1])**2))

def rastrigin(x):
    return float(10*len(x) + np.sum(x*x - 10*np.cos(2*np.pi*x)))

def ackley(x):
    n = len(x)
    return float(-20*np.exp(-0.2*np.sqrt(np.sum(x*x)/n))
                 - np.exp(np.sum(np.cos(2*np.pi*x))/n) + 20 + np.e)

def griewank(x):
    return float(np.sum(x*x)/4000 - np.prod(np.cos(x/np.sqrt(np.arange(1, len(x)+1)))) + 1)

def schwefel(x):
    return float(418.9829*len(x) - np.sum(x*np.sin(np.sqrt(np.abs(x)))))

def levy(x):
    w = 1 + (x-1)/4
    return float(np.sin(np.pi*w[0])**2
                 + np.sum((w[:-1]-1)**2*(1+10*np.sin(np.pi*w[:-1]+1)**2))
                 + (w[-1]-1)**2*(1+np.sin(2*np.pi*w[-1])**2))

def zakharov(x):
    s = np.sum(0.5*np.arange(1, len(x)+1)*x)
    return float(np.sum(x*x) + s**2 + s**4)


FUNCS = {
    "F1 Sphere": (sphere, -100, 100),
    "F2 Rosenbrock": (rosenbrock, -30, 30),
    "F3 Rastrigin": (rastrigin, -5.12, 5.12),
    "F4 Ackley": (ackley, -32, 32),
    "F5 Griewank": (griewank, -600, 600),
    "F6 Schwefel": (schwefel, -500, 500),
    "F7 Levy": (levy, -10, 10),
    "F8 Zakharov": (zakharov, -10, 10),
}


class ObjBudget:
    def __init__(self, f, max_evals=BENCH_MAX_EVALS, grid=BENCH_GRID):
        self.f = f
        self.max_evals = int(max_evals)
        self.evals = 0
        self.best = np.inf
        self.grid = np.asarray(grid, int)
        self.points = []
        self.gi = 0

    @property
    def remaining(self):
        return self.max_evals - self.evals

    def evaluate(self, x):
        if self.evals >= self.max_evals:
            raise RuntimeError("Benchmark evaluation budget exhausted")
        y = float(self.f(x))
        self.evals += 1
        if y < self.best:
            self.best = y
        while self.gi < len(self.grid) and self.evals >= self.grid[self.gi]:
            self.points.append(self.best)
            self.gi += 1
        return y

    def curve(self):
        while len(self.points) < len(self.grid):
            self.points.append(self.best)
        return np.asarray(self.points, float)


def _t0_estimate_counted(budget, x, lo, hi, rng, n=30):
    vals = []
    for _ in range(min(n, budget.remaining)):
        xn = np.clip(x + rng.normal(0, 0.1*(hi-lo), BD), lo, hi)
        vals.append(budget.evaluate(xn))
    return max(1e-12, float(np.std(vals))) if vals else 1e-12


def bench_sa(f, lo, hi, seed, max_evals=BENCH_MAX_EVALS):
    rng = np.random.default_rng(seed)
    b = ObjBudget(f, max_evals)

    x = rng.uniform(lo, hi, BD)
    fx = b.evaluate(x)
    T = _t0_estimate_counted(b, x, lo, hi, rng)
    alpha = (1e-8)**(1.0/max(max_evals, 1))

    while b.remaining > 0:
        xn = np.clip(x + rng.normal(0, 0.1*(hi-lo), BD), lo, hi)
        fn = b.evaluate(xn)
        d = fn-fx
        if d < 0 or rng.random() < np.exp(-d/max(T, 1e-300)):
            x, fx = xn, fn
        T *= alpha

    return b.best, b.curve(), b.evals


def bench_amqsa(f, lo, hi, seed, max_evals=BENCH_MAX_EVALS):
    rng = np.random.default_rng(seed)
    b = ObjBudget(f, max_evals)

    R = hi-lo
    x = rng.uniform(lo, hi, BD)
    fx = b.evaluate(x)
    bx, bf = x.copy(), fx
    T = _t0_estimate_counted(b, x, lo, hi, rng)
    alpha = (1e-8)**(1.0/max(max_evals, 1))
    wts = np.ones(4)
    quenched = False

    while b.remaining > 0:
        if not quenched and b.evals >= QUENCH_FRAC*max_evals:
            T = 0.0
            quenched = True
            x, fx = bx.copy(), bf

        pr = np.maximum(wts, 0.05)
        pr = pr/pr.sum()
        oi = rng.choice(4, p=pr)

        if oi == 0:
            xn = x + rng.normal(0, 0.10*R, BD)
        elif oi == 1:
            xn = x + rng.normal(0, 0.01*R, BD)
        elif oi == 2:
            xn = x.copy()
            k = rng.integers(0, BD)
            xn[k] += rng.normal(0, 0.10*R)
        else:
            xn = x + 0.05*R*rng.standard_cauchy(BD)

        xn = np.clip(xn, lo, hi)
        fn = b.evaluate(xn)
        d = fn-fx
        acc = d < 0 or (T > 0 and rng.random() < np.exp(-d/max(T, 1e-300)))
        wts[oi] = 0.9*wts[oi] + 0.1*(1.0 if d < 0 else (0.3 if acc else 0.0))

        if acc:
            x, fx = xn, fn
            if fn < bf:
                bx, bf = xn.copy(), fn

        T *= alpha

    return b.best, b.curve(), b.evals


def bench_ga(f, lo, hi, seed, max_evals=BENCH_MAX_EVALS):
    rng = np.random.default_rng(seed)
    b = ObjBudget(f, max_evals)

    pop = rng.uniform(lo, hi, (BPOP, BD))
    fits = np.array([b.evaluate(p) for p in pop])

    while b.remaining >= BPOP:
        new = []
        for _ in range(BPOP):
            i1 = min(rng.choice(BPOP, 3, replace=False), key=lambda i: fits[i])
            i2 = min(rng.choice(BPOP, 3, replace=False), key=lambda i: fits[i])
            beta = rng.uniform(-0.25, 1.25, BD)
            child = beta*pop[i1] + (1-beta)*pop[i2]
            mmask = rng.random(BD) < 0.1
            child = child + mmask*rng.normal(0, 0.05*(hi-lo), BD)
            new.append(np.clip(child, lo, hi))

        pop = np.array(new)
        fits = np.array([b.evaluate(p) for p in pop])

    # BENCH_MAX_EVALS=10,000 and BPOP=30 leaves 10 evaluations after full batches.
    # Use the remainder on independently generated offspring so the total is exact.
    while b.remaining > 0:
        i1 = min(rng.choice(BPOP, 3, replace=False), key=lambda i: fits[i])
        i2 = min(rng.choice(BPOP, 3, replace=False), key=lambda i: fits[i])
        beta = rng.uniform(-0.25, 1.25, BD)
        child = beta*pop[i1] + (1-beta)*pop[i2]
        child = np.clip(child, lo, hi)
        b.evaluate(child)

    return b.best, b.curve(), b.evals


def bench_pso(f, lo, hi, seed, max_evals=BENCH_MAX_EVALS):
    rng = np.random.default_rng(seed)
    b = ObjBudget(f, max_evals)

    x = rng.uniform(lo, hi, (BPOP, BD))
    v = np.zeros_like(x)
    fits = np.array([b.evaluate(p) for p in x])

    pb, pf = x.copy(), fits.copy()
    gb = x[fits.argmin()].copy()

    it = 0
    approx_iters = max(1, math.ceil((max_evals-BPOP)/BPOP))

    while b.remaining > 0:
        w = 0.9 - 0.5*min(it/approx_iters, 1.0)
        r1, r2 = rng.random((BPOP, BD)), rng.random((BPOP, BD))
        v = np.clip(
            w*v + 1.8*r1*(pb-x) + 1.8*r2*(gb-x),
            -0.2*(hi-lo), 0.2*(hi-lo)
        )
        x = np.clip(x+v, lo, hi)

        m = min(BPOP, b.remaining)
        for i in range(m):
            fi = b.evaluate(x[i])
            if fi < pf[i]:
                pb[i], pf[i] = x[i].copy(), fi

        gb = pb[pf.argmin()].copy()
        it += 1

    return b.best, b.curve(), b.evals


def bench_zoa(f, lo, hi, seed, max_evals=BENCH_MAX_EVALS):
    rng = np.random.default_rng(seed)
    b = ObjBudget(f, max_evals)

    x = rng.uniform(lo, hi, (BPOP, BD))
    fits = np.array([b.evaluate(p) for p in x])
    pz = x[fits.argmin()].copy()
    pzf = fits.min()

    while b.remaining > 0:
        for i in range(BPOP):
            if b.remaining <= 0:
                break

            r = rng.random(BD)
            I = 1 + rng.integers(0, 2)
            xn = np.clip(x[i] + r*(pz-I*x[i]), lo, hi)
            fn = b.evaluate(xn)
            if fn < fits[i]:
                x[i], fits[i] = xn, fn
            if fits[i] < pzf:
                pz, pzf = x[i].copy(), fits[i]

            if b.remaining <= 0:
                break

            progress = b.evals/max_evals
            if rng.random() < 0.5:
                xn = np.clip(
                    x[i] + 0.01*(2*rng.random(BD)-1)*(1-progress)*x[i],
                    lo, hi
                )
            else:
                k = rng.integers(0, BPOP)
                r = rng.random(BD)
                I = 1 + rng.integers(0, 2)
                xn = np.clip(x[i] + r*(x[k]-I*x[i]), lo, hi)

            fn = b.evaluate(xn)
            if fn < fits[i]:
                x[i], fits[i] = xn, fn
            if fits[i] < pzf:
                pz, pzf = x[i].copy(), fits[i]

    return b.best, b.curve(), b.evals


BENCH_ALGOS = {
    "SA": bench_sa,
    "GA": bench_ga,
    "PSO": bench_pso,
    "ZOA": bench_zoa,
    "AMQSA": bench_amqsa,
}


# -------------------------- STATISTICS UTILITIES --------------------------
def bootstrap_mean_ci(x, level=0.95, B=5000, seed=12345):
    x = np.asarray(x, float)
    rng = np.random.default_rng(seed)
    means = np.empty(B)
    n = len(x)
    for i in range(B):
        means[i] = rng.choice(x, size=n, replace=True).mean()
    a = (1-level)/2
    return np.quantile(means, [a, 1-a])


def holm_adjust(pvals):
    pvals = np.asarray(pvals, float)
    m = len(pvals)
    order = np.argsort(pvals)
    adj_sorted = np.empty(m)
    running = 0.0
    for k, idx in enumerate(order):
        val = (m-k)*pvals[idx]
        running = max(running, val)
        adj_sorted[k] = min(1.0, running)
    out = np.empty(m)
    for k, idx in enumerate(order):
        out[idx] = adj_sorted[k]
    return out


def paired_rank_biserial(reference, comparator):
    # Positive = reference is better for minimization.
    d = np.asarray(comparator, float) - np.asarray(reference, float)
    d = d[d != 0]
    if len(d) == 0:
        return 0.0
    r = rankdata(np.abs(d))
    wp = r[d > 0].sum()
    wm = r[d < 0].sum()
    return float((wp-wm)/(wp+wm))


print("Definitions loaded.")
print(f"Instance: {N} flights | {DATA_SOURCE}")
print(f"CGK fair budget: {MAX_EVALS:,} total objective evaluations / run")
print(f"Benchmark fair budget: {BENCH_MAX_EVALS:,} total objective evaluations / run")
````

````python

# STEP B4 — Run Phase 2 experiments and save all evidence
# Expected runtime can be substantial on Colab CPU.

from pathlib import Path
import json, time, zipfile

PHASE2_START = time.perf_counter()

config = {
    "cgk_max_evals": MAX_EVALS,
    "cgk_population_size": POP,
    "cgk_runs": N_RUNS,
    "cgk_seeds": list(range(2000, 2000+N_RUNS)),
    "benchmark_max_evals": BENCH_MAX_EVALS,
    "benchmark_population_size": BPOP,
    "benchmark_dimension": BD,
    "benchmark_runs": BN_RUNS,
    "benchmark_seeds": list(range(3000, 3000+BN_RUNS)),
    "quench_fraction": QUENCH_FRAC,
    "data_source": DATA_SOURCE,
    "use_opensky_file": USE_OPENSKY_FILE,
    "normalization_reference": F_REF.tolist(),
    "scalar_weights": W.tolist(),
}
with open(OUTPUT_DIR / "phase2_configuration.json", "w") as f:
    json.dump(config, f, indent=2)


# ============================== CGK =================================
cgk_results = {}
cgk_curves = {}
cgk_best = {}
cgk_rows = []
budget_rows = []

if RUN_CGK:
    print("\n=== PHASE 2A: FAIR-BUDGET CGK COMPARISON ===")
    print(f"Algorithms: {list(ALGOS)}")
    print(f"Runs/algorithm: {N_RUNS}; total evals/run: {MAX_EVALS}")

    for name, fn in ALGOS.items():
        vals = []
        curves = []
        best_global = None

        for r in range(N_RUNS):
            seed = 2000+r
            t0 = time.perf_counter()
            best, curve, nevals = fn(seed=seed, max_evals=MAX_EVALS)
            elapsed = time.perf_counter()-t0

            p, c, (scalar, fvec, ncan) = best
            vals.append(fvec)
            curves.append(curve)

            if best_global is None or scalar < best_global[2][0]:
                best_global = best

            cgk_rows.append({
                "Algorithm": name,
                "Run": r+1,
                "Seed": seed,
                "ScalarFitness": scalar,
                "f1_IDR": fvec[0],
                "f2_IDR": fvec[1],
                "f3_min": fvec[2],
                "Cancelled": ncan,
                "Evaluations": nevals,
                "Runtime_s": elapsed,
            })
            budget_rows.append({
                "Experiment": "CGK",
                "Algorithm": name,
                "Run": r+1,
                "Seed": seed,
                "TargetEvaluations": MAX_EVALS,
                "ActualEvaluations": nevals,
                "BudgetOK": nevals == MAX_EVALS,
            })

        cgk_results[name] = np.asarray(vals)
        cgk_curves[name] = np.vstack(curves)
        cgk_best[name] = best_global

        print(
            f"{name:6s} | f1 mean={cgk_results[name][:,0].mean()/1e9:.6f}e9 "
            f"| evals={MAX_EVALS} | runtime mean="
            f"{np.mean([x['Runtime_s'] for x in cgk_rows if x['Algorithm']==name]):.2f}s"
        )

    cgk_raw = pd.DataFrame(cgk_rows)
    cgk_raw.to_csv(OUTPUT_DIR / "phase2_cgk_raw.csv", index=False)

    # Summary with bootstrap CI
    summary_rows = []
    for name in ALGOS:
        arr = cgk_results[name]
        rtimes = cgk_raw.loc[cgk_raw["Algorithm"]==name, "Runtime_s"].to_numpy()

        row = {"Algorithm": name}
        for j, metric in enumerate(["f1_IDR", "f2_IDR", "f3_min"]):
            x = arr[:,j]
            lo, hi = bootstrap_mean_ci(x, seed=12000+j)
            row.update({
                f"{metric}_mean": x.mean(),
                f"{metric}_std": x.std(ddof=1),
                f"{metric}_median": np.median(x),
                f"{metric}_best": x.min(),
                f"{metric}_worst": x.max(),
                f"{metric}_CI95_low": lo,
                f"{metric}_CI95_high": hi,
            })
        row["Runtime_s_mean"] = rtimes.mean()
        row["Runtime_s_std"] = rtimes.std(ddof=1)
        row["Evaluations_per_run"] = MAX_EVALS
        summary_rows.append(row)

    cgk_summary = pd.DataFrame(summary_rows)
    cgk_summary.to_csv(OUTPUT_DIR / "phase2_cgk_summary.csv", index=False)

    print("\nCGK summary (f1 in billion IDR):")
    display(
        cgk_summary[
            ["Algorithm","f1_IDR_mean","f1_IDR_std","f1_IDR_median",
             "f1_IDR_CI95_low","f1_IDR_CI95_high","Runtime_s_mean"]
        ].assign(
            f1_IDR_mean=lambda d: d.f1_IDR_mean/1e9,
            f1_IDR_std=lambda d: d.f1_IDR_std/1e9,
            f1_IDR_median=lambda d: d.f1_IDR_median/1e9,
            f1_IDR_CI95_low=lambda d: d.f1_IDR_CI95_low/1e9,
            f1_IDR_CI95_high=lambda d: d.f1_IDR_CI95_high/1e9,
        ).round(6)
    )

    # Paired Wilcoxon + Holm + paired rank-biserial effect size
    comparisons = [x for x in ALGOS if x != "AMQSA"]
    stat_rows = []
    raw_ps = []

    ref = cgk_results["AMQSA"][:,0]

    for other in comparisons:
        x = cgk_results[other][:,0]
        wstat, p = wilcoxon(ref, x, alternative="two-sided", zero_method="wilcox")
        raw_ps.append(p)

        stat_rows.append({
            "Reference": "AMQSA",
            "Comparator": other,
            "WilcoxonStatistic": wstat,
            "p_raw": p,
            "RankBiserial_positive_means_AMQSA_better": paired_rank_biserial(ref, x),
            "MeanImprovementPct_positive_means_AMQSA_better":
                (x.mean()-ref.mean())/x.mean()*100.0,
            "AMQSA_f1_mean_IDR": ref.mean(),
            "Comparator_f1_mean_IDR": x.mean(),
            "n_pairs": len(ref),
        })

    adj = holm_adjust(raw_ps)
    for row, p_adj in zip(stat_rows, adj):
        row["p_Holm"] = p_adj

    pairwise = pd.DataFrame(stat_rows)
    pairwise.to_csv(OUTPUT_DIR / "phase2_cgk_pairwise_stats.csv", index=False)

    print("\nPaired Wilcoxon + Holm correction:")
    display(pairwise.round(6))

    # Convergence against OBJECTIVE EVALUATIONS
    fig, ax = plt.subplots(figsize=(8.5, 5))
    for name in ALGOS:
        C = cgk_curves[name]/1e9
        ax.plot(CGK_GRID, C.mean(axis=0), label=name)
        ax.fill_between(
            CGK_GRID,
            C.mean(axis=0)-C.std(axis=0),
            C.mean(axis=0)+C.std(axis=0),
            alpha=0.12
        )
    ax.set_xlabel("Objective-function evaluations")
    ax.set_ylabel("Airline delay cost f1 / 1e9 IDR")
    ax.set_title("CGK convergence under equal evaluation budget")
    ax.grid(alpha=0.3)
    ax.legend(ncol=3)
    plt.tight_layout()
    fig.savefig(OUTPUT_DIR / "phase2_cgk_convergence_equal_budget.png", dpi=220, bbox_inches="tight")
    plt.show()


# ============================ ABLATION ==============================
if RUN_ABLATION:
    print("\n=== PHASE 2B: FAIR-BUDGET ABLATION ===")

    configs = {
        "SA+reheat": dict(ops=("swap","ins","bit"), w0=(1.5,1.5,0.3), reheat=True),
        "M3+oropt": dict(ops=("swap","ins","blk","bit"), w0=(1.5,1.5,1.0,0.3)),
        "M3+pri": dict(ops=("swap","ins","pri","bit"), w0=(1.5,1.5,1.0,0.3)),
    }

    ab_rows = []

    for name, kw in configs.items():
        for r in range(N_RUNS):
            seed = 2000+r
            t0 = time.perf_counter()
            f1, nevals = _ablation_variant(seed, max_evals=MAX_EVALS, **kw)
            elapsed = time.perf_counter()-t0
            ab_rows.append({
                "Configuration": name, "Run": r+1, "Seed": seed,
                "f1_IDR": f1, "Evaluations": nevals, "Runtime_s": elapsed
            })
            budget_rows.append({
                "Experiment": "Ablation", "Algorithm": name, "Run": r+1,
                "Seed": seed, "TargetEvaluations": MAX_EVALS,
                "ActualEvaluations": nevals, "BudgetOK": nevals == MAX_EVALS,
            })

    for r in range(N_RUNS):
        seed = 2000+r
        t0 = time.perf_counter()
        f1, nevals = _quench_only(seed, max_evals=MAX_EVALS)
        elapsed = time.perf_counter()-t0
        ab_rows.append({
            "Configuration": "M3_quench_only", "Run": r+1, "Seed": seed,
            "f1_IDR": f1, "Evaluations": nevals, "Runtime_s": elapsed
        })
        budget_rows.append({
            "Experiment": "Ablation", "Algorithm": "M3_quench_only", "Run": r+1,
            "Seed": seed, "TargetEvaluations": MAX_EVALS,
            "ActualEvaluations": nevals, "BudgetOK": nevals == MAX_EVALS,
        })

    # Reuse exactly the CGK SA and AMQSA runs so the ablation table is internally consistent.
    if RUN_CGK:
        for name in ["SA", "AMQSA"]:
            for _, rr in cgk_raw[cgk_raw.Algorithm==name].iterrows():
                ab_rows.append({
                    "Configuration": name,
                    "Run": int(rr.Run),
                    "Seed": int(rr.Seed),
                    "f1_IDR": rr.f1_IDR,
                    "Evaluations": int(rr.Evaluations),
                    "Runtime_s": rr.Runtime_s,
                })

    ab_raw = pd.DataFrame(ab_rows)
    ab_raw.to_csv(OUTPUT_DIR / "phase2_ablation_raw.csv", index=False)

    ab_summary_rows = []
    for name, g in ab_raw.groupby("Configuration", sort=False):
        x = g.f1_IDR.to_numpy()
        lo, hi = bootstrap_mean_ci(x, seed=15000)
        ab_summary_rows.append({
            "Configuration": name,
            "f1_mean_IDR": x.mean(),
            "f1_std_IDR": x.std(ddof=1),
            "f1_median_IDR": np.median(x),
            "f1_CI95_low": lo,
            "f1_CI95_high": hi,
            "Evaluations_per_run": int(g.Evaluations.iloc[0]),
            "Runtime_s_mean": g.Runtime_s.mean(),
        })

    ab_summary = pd.DataFrame(ab_summary_rows)

    if "SA" in set(ab_summary.Configuration):
        sa_mean = float(ab_summary.loc[ab_summary.Configuration=="SA","f1_mean_IDR"].iloc[0])
        ab_summary["Improvement_vs_SA_pct"] = (
            (sa_mean-ab_summary["f1_mean_IDR"])/sa_mean*100.0
        )

    ab_summary.to_csv(OUTPUT_DIR / "phase2_ablation_summary.csv", index=False)

    print("\nAblation summary:")
    show = ab_summary.copy()
    show["f1_mean_billion_IDR"] = show.f1_mean_IDR/1e9
    display(show[["Configuration","f1_mean_billion_IDR","Improvement_vs_SA_pct",
                  "Evaluations_per_run","Runtime_s_mean"]].round(5))


# =========================== BENCHMARK ==============================
bench_rows = []
bench_curves = {}

if RUN_BENCHMARK:
    print("\n=== PHASE 2C: FAIR-BUDGET CONTINUOUS BENCHMARK ===")
    print(
        "Important: SA/AMQSA temperature-estimation evaluations are INCLUDED "
        "inside the 10,000-evaluation budget."
    )

    for fname, (f, lo, hi) in FUNCS.items():
        bench_curves[fname] = {}

        for aname, algo in BENCH_ALGOS.items():
            curves = []

            for r in range(BN_RUNS):
                seed = 3000+r
                t0 = time.perf_counter()
                best, curve, nevals = algo(f, lo, hi, seed, BENCH_MAX_EVALS)
                elapsed = time.perf_counter()-t0

                curves.append(curve)
                bench_rows.append({
                    "Function": fname,
                    "Algorithm": aname,
                    "Run": r+1,
                    "Seed": seed,
                    "BestObjective": best,
                    "Evaluations": nevals,
                    "Runtime_s": elapsed,
                })
                budget_rows.append({
                    "Experiment": f"Benchmark::{fname}",
                    "Algorithm": aname,
                    "Run": r+1,
                    "Seed": seed,
                    "TargetEvaluations": BENCH_MAX_EVALS,
                    "ActualEvaluations": nevals,
                    "BudgetOK": nevals == BENCH_MAX_EVALS,
                })

            bench_curves[fname][aname] = np.vstack(curves)

        means = {
            a: np.mean([x["BestObjective"] for x in bench_rows
                       if x["Function"]==fname and x["Algorithm"]==a])
            for a in BENCH_ALGOS
        }
        print(fname, "|", " | ".join(f"{a}={means[a]:.4e}" for a in BENCH_ALGOS))

    bench_raw = pd.DataFrame(bench_rows)
    bench_raw.to_csv(OUTPUT_DIR / "phase2_benchmark_raw.csv", index=False)

    # Descriptive benchmark summary
    bs_rows = []
    for (fname, aname), g in bench_raw.groupby(["Function","Algorithm"], sort=False):
        x = g.BestObjective.to_numpy()
        lo, hi = bootstrap_mean_ci(x, seed=17000)
        bs_rows.append({
            "Function": fname,
            "Algorithm": aname,
            "mean": x.mean(),
            "std": x.std(ddof=1),
            "median": np.median(x),
            "best": x.min(),
            "worst": x.max(),
            "CI95_low": lo,
            "CI95_high": hi,
            "Runtime_s_mean": g.Runtime_s.mean(),
            "Evaluations_per_run": int(g.Evaluations.iloc[0]),
        })

    bench_summary = pd.DataFrame(bs_rows)
    bench_summary.to_csv(OUTPUT_DIR / "phase2_benchmark_summary.csv", index=False)

    # IMPORTANT statistical unit: one benchmark FUNCTION = one Friedman block.
    alg_order = list(BENCH_ALGOS)
    func_order = list(FUNCS)
    mean_matrix = np.array([
        [
            bench_raw[
                (bench_raw.Function==fname) &
                (bench_raw.Algorithm==a)
            ].BestObjective.mean()
            for a in alg_order
        ]
        for fname in func_order
    ])

    fried_stat, fried_p = friedmanchisquare(
        *[mean_matrix[:,j] for j in range(len(alg_order))]
    )

    function_ranks = np.vstack([
        rankdata(row, method="average")
        for row in mean_matrix
    ])
    avg_ranks = function_ranks.mean(axis=0)

    fried_df = pd.DataFrame({
        "Algorithm": alg_order,
        "AverageRank_across_8_functions": avg_ranks,
    })
    fried_df["Friedman_chi2"] = fried_stat
    fried_df["Friedman_p"] = fried_p
    fried_df["StatisticalUnit"] = "benchmark function (n=8), using 20-run mean per function"
    fried_df.to_csv(OUTPUT_DIR / "phase2_benchmark_friedman.csv", index=False)

    print("\nBenchmark Friedman test:")
    print(f"chi2={fried_stat:.6f}, p={fried_p:.6g}")
    display(fried_df)

    # 8 panels: convergence vs objective evaluations
    fig, axes = plt.subplots(2, 4, figsize=(15, 7))
    for ax, fname in zip(axes.ravel(), FUNCS):
        for a in BENCH_ALGOS:
            C = bench_curves[fname][a]
            ax.plot(BENCH_GRID, np.maximum(C.mean(axis=0), 1e-30), label=a)
        ax.set_yscale("log")
        ax.set_title(fname, fontsize=9)
        ax.set_xlabel("evaluations", fontsize=8)
        ax.grid(alpha=0.3)
    axes[0,0].legend(fontsize=7)
    plt.tight_layout()
    fig.savefig(OUTPUT_DIR / "phase2_benchmark_convergence_equal_budget.png",
                dpi=220, bbox_inches="tight")
    plt.show()


# =========================== BUDGET PROOF ===========================
budget_df = pd.DataFrame(budget_rows)
budget_df.to_csv(OUTPUT_DIR / "phase2_budget_check.csv", index=False)

budget_summary = (
    budget_df.groupby(["Experiment","Algorithm"], as_index=False)
    .agg(
        TargetEvaluations=("TargetEvaluations","first"),
        MinActualEvaluations=("ActualEvaluations","min"),
        MaxActualEvaluations=("ActualEvaluations","max"),
        AllRunsBudgetOK=("BudgetOK","all"),
        Runs=("Run","count"),
    )
)
budget_summary.to_csv(OUTPUT_DIR / "phase2_budget_summary.csv", index=False)

print("\n=== BUDGET VERIFICATION ===")
display(budget_summary)

if not budget_df["BudgetOK"].all():
    raise RuntimeError("STOP: at least one run violated the declared evaluation budget.")

print("\nALL RUNS PASSED THE EQUAL-EVALUATION-BUDGET CHECK.")


# ============================ ZIP OUTPUT ============================
elapsed_total = time.perf_counter()-PHASE2_START

with open(OUTPUT_DIR / "phase2_completion.json", "w") as f:
    json.dump({
        "completed": True,
        "elapsed_seconds": elapsed_total,
        "all_budget_checks_passed": bool(budget_df["BudgetOK"].all()),
        "cgk_target_evals": MAX_EVALS,
        "benchmark_target_evals": BENCH_MAX_EVALS,
    }, f, indent=2)

zip_path = Path("/content/AMQSA_PHASE2_FAIR_BUDGET_RESULTS.zip")
with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as z:
    for p in OUTPUT_DIR.rglob("*"):
        if p.is_file():
            z.write(p, arcname=p.relative_to(OUTPUT_DIR))

print(f"\nPHASE 2 COMPLETE in {elapsed_total/60:.1f} minutes.")
print("Results folder:", OUTPUT_DIR)
print("ZIP file:", zip_path)
print("\nDownload the ZIP and this executed notebook, then upload both back to ChatGPT.")
````
